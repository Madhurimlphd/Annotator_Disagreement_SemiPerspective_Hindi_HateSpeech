**In this code, we calculated the disagreement between original tweets and relabelled tweets for HASOC2020**

```python
import pandas as pd
import numpy as np
```

```python
import os

print(os.getcwd())
print(os.listdir("dataset_reannotate"))
```

```python

train_df = pd.read_excel("dataset_reannotate/Hasoc2020_reannotated.xlsx")
train_df= train_df.dropna(how='all')
```

```python
train_df = train_df.drop(columns=["task2"])
```

```python
len(train_df)
```

```python
train_df.head(20)
```

```python
train_df.Relabel.value_counts()
```

Filling the empty value in New_label column

```python
train_df["Relabel"] = train_df["Relabel"].fillna(train_df["task1"])
```

Finding the Disagreement Tweets

```python
train_df['Mismatch'] = (train_df['task1'] != train_df['Relabel']).astype(int)
```

```python
mismatch_count = (train_df["Mismatch"] == 1).sum()
print("Count of tweets with mismatch column having value 1 indicating difference in original and new label",mismatch_count)
```

```python
train_df.to_csv("dataset_meta/hindi2020_ds.csv")
```

**Cohen Kappa between original label and relabelled**

```python
from sklearn.metrics import cohen_kappa_score
score=cohen_kappa_score(train_df["task1"],train_df["Relabel"])
print("Cohen Kappa between original and relabel for HASOC 2020:",score)
```
